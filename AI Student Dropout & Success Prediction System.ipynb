```python
import pandas as pd

df = x.copy()
df["Target"] = y.iloc[:, 0].values

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
display(df.head())
```

```python
X = df.drop("Target", axis=1)
y = df["Target"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)

print("\nTarget classes:")
print(y.unique())
```

```python
categorical_columns = [
    "Marital Status",
    "Application mode",
    "Course",
    "Daytime/evening attendance",
    "Previous qualification",
    "Nacionality",
    "Mother's qualification",
    "Father's qualification",
    "Mother's occupation",
    "Father's occupation",
    "Displaced",
    "Educational special needs",
    "Debtor",
    "Tuition fees up to date",
    "Gender",
    "Scholarship holder",
    "International"
]

numerical_columns = [
    column for column in X.columns
    if column not in categorical_columns
]

print("Categorical variables:", len(categorical_columns))
print("Numerical variables:", len(numerical_columns))

print("\nCategorical variables:")
print(categorical_columns)

print("\nNumerical variables:")
print(numerical_columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTesting set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        ),
        (
            "numerical",
            StandardScaler(),
            numerical_columns
        )
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Processed training data shape:", X_train_processed.shape)
print("Processed testing data shape:", X_test_processed.shape)
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Target encoding:")
for number, class_name in enumerate(label_encoder.classes_):
    print(number, "=", class_name)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Logistic Regression model
logistic_model = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
logistic_model.fit(X_train_processed, y_train_encoded)

print("Logistic Regression model trained successfully.")
```

```python
# Predict the test data
y_pred_logistic = logistic_model.predict(X_test_processed)

print("First 10 predicted values:")
print(y_pred_logistic[:10])

print("\nFirst 10 actual values:")
print(y_test_encoded[:10])
```

```python
logistic_accuracy = accuracy_score(y_test_encoded, y_pred_logistic)

print("Logistic Regression Accuracy:", logistic_accuracy)
```

```python
print("Logistic Regression Classification Report:")
print(
    classification_report(
        y_test_encoded,
        y_pred_logistic,
        target_names=label_encoder.classes_
    )
)
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

# Train the model
random_forest_model.fit(
    X_train_processed,
    y_train_encoded
)

print("Random Forest model trained successfully.")
```

```python
# Predict the test data
y_pred_rf = random_forest_model.predict(X_test_processed)

print("First 10 predicted values:")
print(y_pred_rf[:10])

print("\nFirst 10 actual values:")
print(y_test_encoded[:10])
```

```python
rf_accuracy = accuracy_score(
    y_test_encoded,
    y_pred_rf
)

print("Random Forest Accuracy:", rf_accuracy)
print("Random Forest Accuracy (%):", rf_accuracy * 100)
```

```python
print("Random Forest Classification Report:")
print(
    classification_report(
        y_test_encoded,
        y_pred_rf,
        target_names=label_encoder.classes_
    )
)
```

```python
rf_cm = confusion_matrix(
    y_test_encoded,
    y_pred_rf
)

print("Random Forest Confusion Matrix:")
print(rf_cm)
```

```python
print("MODEL COMPARISON")
print("=" * 40)

print(f"Logistic Regression Accuracy: {logistic_accuracy:.4f}")
print(f"Random Forest Accuracy:       {rf_accuracy:.4f}")
```

```python
print("\nLogistic Regression:")
print(
    classification_report(
        y_test_encoded,
        y_pred_logistic,
        target_names=label_encoder.classes_
    )
)

print("\nRandom Forest:")
print(
    classification_report(
        y_test_encoded,
        y_pred_rf,
        target_names=label_encoder.classes_
    )
)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

model_results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest",
        "XGBoost"
    ],
    "Accuracy": [
        accuracy_score(y_test_encoded, y_pred_logistic),
        accuracy_score(y_test_encoded, y_pred_rf),
        accuracy_score(y_test_encoded, y_pred_xgb)
    ],
    "Macro Precision": [
        precision_score(y_test_encoded, y_pred_logistic, average="macro"),
        precision_score(y_test_encoded, y_pred_rf, average="macro"),
        precision_score(y_test_encoded, y_pred_xgb, average="macro")
    ],
    "Macro Recall": [
        recall_score(y_test_encoded, y_pred_logistic, average="macro"),
        recall_score(y_test_encoded, y_pred_rf, average="macro"),
        recall_score(y_test_encoded, y_pred_xgb, average="macro")
    ],
    "Macro F1": [
        f1_score(y_test_encoded, y_pred_logistic, average="macro"),
        f1_score(y_test_encoded, y_pred_rf, average="macro"),
        f1_score(y_test_encoded, y_pred_xgb, average="macro")
    ]
})

print("Complete Model Evaluation:")
display(model_results.round(4))
```
