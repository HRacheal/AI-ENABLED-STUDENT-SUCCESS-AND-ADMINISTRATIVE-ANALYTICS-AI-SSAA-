```python
import pandas as pd
import numpy as np

# Set random seed so the same dataset is generated every time
np.random.seed(42)

# Number of administrative requests
n = 10000

# --------------------------------------------------
# 1. Generate Request IDs
# --------------------------------------------------

request_id = [
    f"ADM{i:05d}"
    for i in range(1, n + 1)
]


# --------------------------------------------------
# 2. Generate Departments
# --------------------------------------------------

departments = np.random.choice(
    [
        "Registrar",
        "Finance",
        "Admissions",
        "IT",
        "Library",
        "Student Affairs",
        "Academic Affairs"
    ],
    size=n,
    p=[
        0.22,
        0.18,
        0.15,
        0.18,
        0.08,
        0.10,
        0.09
    ]
)


# --------------------------------------------------
# 3. Generate Request Types
# --------------------------------------------------

request_types = np.random.choice(
    [
        "Course Registration",
        "Transcript Request",
        "Fee Statement",
        "Payment Issue",
        "Application Issue",
        "Portal Access",
        "Password Reset",
        "Library Issue",
        "Accommodation Request",
        "Student ID Issue",
        "Exam or Assessment Issue",
        "Graduation Clearance"
    ],
    size=n
)


# --------------------------------------------------
# 4. Generate Priority
# --------------------------------------------------

priorities = np.random.choice(
    [
        "Low",
        "Medium",
        "High",
        "Critical"
    ],
    size=n,
    p=[
        0.20,
        0.45,
        0.30,
        0.05
    ]
)


# --------------------------------------------------
# 5. Generate Communication Channel
# --------------------------------------------------

channels = np.random.choice(
    [
        "Email",
        "Portal",
        "Phone",
        "Walk-in",
        "Helpdesk"
    ],
    size=n,
    p=[
        0.30,
        0.30,
        0.15,
        0.10,
        0.15
    ]
)


# --------------------------------------------------
# 6. Generate First Response Time
#    Higher priority = generally faster response
# --------------------------------------------------

response_time = []

for priority in priorities:

    if priority == "Critical":
        time = np.random.uniform(0.2, 2)

    elif priority == "High":
        time = np.random.uniform(0.5, 5)

    elif priority == "Medium":
        time = np.random.uniform(2, 10)

    else:
        time = np.random.uniform(4, 24)

    response_time.append(time)


response_time = np.round(response_time, 2)


# --------------------------------------------------
# 7. Generate Number of Interactions
# --------------------------------------------------

interactions = np.random.poisson(
    lam=3,
    size=n
) + 1


# --------------------------------------------------
# 8. Generate Escalation
#    0 = No
#    1 = Yes
# --------------------------------------------------

escalated = np.random.choice(
    [0, 1],
    size=n,
    p=[0.80, 0.20]
)


# --------------------------------------------------
# 9. Define Base Resolution Time
#    Different request types naturally take
#    different amounts of time
# --------------------------------------------------

base_time = {
    "Password Reset": 2,
    "Portal Access": 4,
    "Student ID Issue": 6,
    "Library Issue": 8,
    "Course Registration": 12,
    "Transcript Request": 15,
    "Fee Statement": 18,
    "Application Issue": 20,
    "Accommodation Request": 24,
    "Exam or Assessment Issue": 22,
    "Graduation Clearance": 28,
    "Payment Issue": 30
}


# --------------------------------------------------
# 10. Generate Resolution Time
# --------------------------------------------------

resolution_time = []

for i in range(n):

    # Base time based on request type
    base = base_time[request_types[i]]

    # First response contributes to total resolution time
    response_effect = response_time[i]

    # More interactions generally mean more time
    interaction_effect = interactions[i] * 1.5

    # Escalated requests generally take longer
    escalation_effect = escalated[i] * 10

    # Random variation makes the data more realistic
    random_effect = np.random.normal(0, 3)

    # Calculate total resolution time
    total_time = (
        base
        + response_effect
        + interaction_effect
        + escalation_effect
        + random_effect
    )

    # Prevent unrealistic negative values
    total_time = max(total_time, 0.5)

    resolution_time.append(total_time)


resolution_time = np.round(resolution_time, 2)


# --------------------------------------------------
# 11. Generate Satisfaction Score
#    Faster resolution generally gives higher
#    satisfaction
# --------------------------------------------------

satisfaction = []

for time in resolution_time:

    if time <= 8:
        score = np.random.choice([4, 5])

    elif time <= 24:
        score = np.random.choice([3, 4, 5])

    elif time <= 48:
        score = np.random.choice([2, 3, 4])

    else:
        score = np.random.choice([1, 2, 3])

    satisfaction.append(score)


# --------------------------------------------------
# 12. Generate Status
# --------------------------------------------------

status = []

for i in range(n):

    if escalated[i] == 1:
        request_status = np.random.choice(
            ["Escalated", "Resolved"],
            p=[0.30, 0.70]
        )

    else:
        request_status = np.random.choice(
            ["Resolved", "Pending"],
            p=[0.90, 0.10]
        )

    status.append(request_status)


# --------------------------------------------------
# 13. Create the Final DataFrame
# --------------------------------------------------

df = pd.DataFrame({
    "Request_ID": request_id,
    "Department": departments,
    "Request_Type": request_types,
    "Priority": priorities,
    "Channel": channels,
    "First_Response_Time": response_time,
    "Number_of_Interactions": interactions,
    "Escalated": escalated,
    "Resolution_Time": resolution_time,
    "Satisfaction_Score": satisfaction,
    "Status": status
})


# --------------------------------------------------
# 14. Display Dataset Information
# --------------------------------------------------

print("Synthetic University Administrative Dataset")
print("=" * 50)

print("\nDataset shape:")
print(df.shape)

print("\nFirst 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())


# --------------------------------------------------
# 15. Save Dataset as CSV
# --------------------------------------------------

file_name = "synthetic_university_administrative_requests.csv"

df.to_csv(
    file_name,
    index=False
)

print("\n" + "=" * 50)
print("Dataset saved successfully!")
print("File name:", file_name)
print("=" * 50)
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# 1. Load the dataset
df = pd.read_csv("synthetic_university_administrative_requests.csv")

# 2. Separate Features (X) and Target (y)
feature_cols = [
    'Department', 'Request_Type', 'Priority', 'Channel', 
    'First_Response_Time', 'Number_of_Interactions', 'Escalated'
]
target_col = 'Resolution_Time'

X = df[feature_cols]
y = df[target_col]

# 3. Perform Train-Test Split (80% Train / 20% Test)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 4. Construct ColumnTransformer Pipeline
categorical_cols = ['Department', 'Request_Type', 'Priority', 'Channel']
numerical_cols = ['First_Response_Time', 'Number_of_Interactions', 'Escalated']

preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols),
        ('num', StandardScaler(), numerical_cols)
    ]
)

# 5. Preprocess Datasets
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print(f"X_train Shape: {X_train_processed.shape}")
print(f"X_test Shape:  {X_test_processed.shape}")
```

```python
import os
import warnings
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Suppress joblib parallel warnings
warnings.filterwarnings("ignore", category=UserWarning)

# --- Part A: Regression Models ---
# 1. Baseline Model: Linear Regression
lr_model = LinearRegression()
lr_model.fit(X_train_processed, y_train)

# 2. Ensemble Model: Random Forest Regressor (n_jobs omitted to prevent Windows subprocess lookup)
rf_model = RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42)
rf_model.fit(X_train_processed, y_train)

# --- Part B: K-Means Clustering ---
clustering_features = ['First_Response_Time', 'Resolution_Time', 'Satisfaction_Score']
cluster_scaler = StandardScaler()
clustering_scaled = cluster_scaler.fit_transform(df[clustering_features])

# Fit 3 Clusters
kmeans_model = KMeans(n_clusters=3, random_state=42, n_init=10)
df['Cluster'] = kmeans_model.fit_predict(clustering_scaled)

print("Models and K-Means clustering successfully trained!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, silhouette_score

# 1. Generate Predictions on Test Data
lr_preds = lr_model.predict(X_test_processed)
rf_preds = rf_model.predict(X_test_processed)

# 2. Evaluate Metrics
def compute_metrics(y_true, y_pred, model_name):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = mean_squared_error(y_true, y_pred) ** 0.5
    r2 = r2_score(y_true, y_pred)
    return {"Model": model_name, "MAE (Hours)": round(mae, 2), "RMSE (Hours)": round(rmse, 2), "R² Score": round(r2, 4)}

evaluation_df = pd.DataFrame([
    compute_metrics(y_test, lr_preds, "Linear Regression"),
    compute_metrics(y_test, rf_preds, "Random Forest Regressor")
])

print("=== Regression Model Performance Comparison ===")
display(evaluation_df)

# 3. Evaluate Clustering Quality
sil_score = silhouette_score(clustering_scaled, df['Cluster'])
print(f"\nK-Means Silhouette Score: {sil_score:.4f}")

print("\n=== Cluster Summary (Averages) ===")
cluster_summary = df.groupby('Cluster')[clustering_features].mean().reset_index()
display(cluster_summary.round(2))
```
